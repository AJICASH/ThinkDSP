```python
from thinkdsp import read_wave

wave = read_wave('bell0001.wav')
wave.normalize()
wave.make_audio()
```

```python
wave.plot()
```

```python
segment = wave.segment(start=5.44, duration=0.06)
segment.make_audio()
```

```python
segment.plot()
```

```python
segment.segment(start=5.46, duration=0.001).plot()
```

```python
spectrum = segment.make_spectrum()
spectrum.plot(high=7000)
```
